## Sentiment Analysis 
The sentiment property returns aa namedtuple of the form Sentiment (polarity, subjectivity). 

The polarity score is a flat from -1 to 1. 

The subjectivity is a float from 0-1 where 0 is objective and 1 is subjective 


```python
testimonial= TextBlob("This is amazingly simple to use. What great fun!")
testimonial.sentiment
```

## WordNet Integration 
You can access the synsets for a Word via the synsets property or the get_synsets methos, optionally passing in a part of speech.

```python
from textblob import Word 
from textblob.wordnet import VERB
word = Word("octopus")
word.synsets
```

```python
Word("hack").get_synsets(pos=VERB)
```

You can access the definitions for each synset via the definitions property or define() method, which can also take an optional part-of-speech argument 

```python
Word("octopus").definitions
```

```python
from textblob.wordnet import Synset 
octopus= Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
octopus.path_similarity(shrimp)
```

## Get Word and Noun Phrase Frequencies 
There are two ways to get the frequency of a word or noun phrase in a TextBlob

The first is through the word_counts dictionary.

```python
monty = TextBlob("We are no longer the Knights who say Ni."
                 "We are now the Knights who say Ekki ekki ekki PTANG.")

monty.word_counts["ekki"]
```

```python
monty.words.count("ekki")
```

```python
monty.words.count("ekki", case_sensitive = True)
```

```python
wiki.noun_phrases.count("python")
```

## n-grams
The TextBlob.ngrams() method returns a list of tuples of n successive words

```python
blob = TextBlob("Now is better than never.")
blob.ngrams(n=3)
```

## Get Start and End Indices 
Use sentence.start and sentence.end to get the indices where a sentence starts and ends within a TextBlob

```python
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```
