## Sentiment Analysis 
The sentiment property returns aa namedtuple of the form Sentiment (polarity, subjectivity). 

The polarity score is a flat from -1 to 1. 

The subjectivity is a float from 0-1 where 0 is objective and 1 is subjective 


```python
testimonial= TextBlob("This is amazingly simple to use. What great fun!")
testimonial.sentiment
```

```python
testimonial.sentiment.polarity
```

## Tokenization 
shows us how to break TextBlobs into words or sentences 

```python
zen = TextBlob("Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex.")
zen.words
```

```python
zen.sentences 
```
